```python
# 4.BCELoss:二分类，多标签分类问题

# 样本1: 真实=1, 预测=0.9 → loss = -log(0.9) = 0.105
# 样本2: 真实=0, 预测=0.1 → loss = -log(0.9) = 0.105
# 样本3: 真实=1, 预测=0.6 → loss = -log(0.6) = 0.511
# 样本4: 真实=0, 预测=0.4 → loss = -log(0.6) = 0.511

# 平均 loss = (0.105+0.105+0.511+0.511)/4 = 0.308
# 公式：- [ y_true × log(y_pred) + (1 - y_true) × log(1 - y_pred) ]
#     = - [ 1 × log(0.9) + 0 × log(0.1) ]
#     = - [ log(0.9) + 0 ]
#     = -log(0.9) = 0.105

true_label = torch.tensor([1.0, 0.0, 1.0, 0.0])  # 真实样本
pred_prob = torch.tensor([0.9, 0.1, 0.6, 0.4])  # 预测样本

# BCELoss
loss = BCELoss()
result = loss(pred_prob, true_label)
print(result)
```

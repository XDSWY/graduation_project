```python
import pandas as pd
import numpy as np
import torch
```

```python

diseases = ['N', 'D', 'G', 'C', 'A', 'H', 'M', 'O']
excel_path = r'C:\Users\lenovo\Desktop\graduation_project\data\raw\ODIR-5K\data.xlsx'  
df = pd.read_excel(excel_path)
```

```python
first_row = df.iloc[0]
print(first_row)
```

```python
left_eye = first_row['Left-Fundus']
right_eye = first_row['Right-Fundus']
print(left_eye)
print(right_eye)
```

```python
labels = []
label_map = {}
for disease in diseases:  
    value = first_row[disease]  
    labels.append(value) 
label_map[first_row['Left-Fundus']] = np.array(labels)
print(label_map)
```

```python
img_name = '0_left.jpg'
labels = label_map[img_name]
print(labels)
```

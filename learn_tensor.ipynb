第0维是按行方向，第1维是按列方向，第2维是按深度方向

```python
X = torch.arange(12,dtype=torch.float32).reshape(3,4)
Y = torch.tensor([[2.0,1,4,3],[1,2,3,4],[4,3,2,1]])
torch.cat((X,Y),dim=0), torch.cat((X,Y),dim=1)   # dim=0代表上下拼接，dim=1代表左右拼接
```

```python
X == Y
```

```python
X.sum()   # 对张量中的所有元素进行求和会产生一个只有一个元素的张量
```

```python
a = torch.arange(3).reshape(3,1)  # 广播机制
b = torch.arange(2).reshape(1,2)
a,b
```

```python
a + b
```

```python
X[-1], X[1:3]   # -1表示选择最后一个元素，[1：3]表示选择第二个和第三个元素
```

```python
X[1,2] = 9   # 改写元素
X
```

```python
X[0:2,:] = 12  # 将一二行复制为12
X
```

```python
before = id(Y)
Y = Y + X
id(Y) == before 
```

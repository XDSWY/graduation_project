```python
import torch
import torch.nn as nn
import pandas as pd
import numpy as np
from tqdm.notebook import tqdm
from torch.utils.data import DataLoader
from torch.utils.tensorboard import SummaryWriter
from src.data.dataset import FundusDataset
from transformers import ViTForImageClassification, ViTImageProcessor
```

```python
# 超参数
config = {
    'model_name': 'google/vit-base-patch16-224',  # 预训练ViT模型
    'num_classes': 8,                              # 分类数
    'class_names': ['正常', '糖尿病视网膜病变', '青光眼', '白内障', 
                   '黄斑变性', '高血压视网膜病变', '近视', '其他'],
    'batch_size': 16,                               # 批次大小(16)
    'epochs': 10,                                   # 训练轮数(30)
    'learning_rate': 2e-5,                          # 学习率
     'weight_decay': 0.01,                           # 权重衰减
     'warmup_steps': 500,                            # 预热步数
     'gradient_accumulation_steps': 2,               # 梯度累积
    'max_grad_norm': 1.0,                           # 梯度裁剪
     'save_dir': './checkpoints',                    # 模型保存目录
    'log_dir': './logs'                              # TensorBoard日志目录
 }
```

```python
def cpu():
    if torch.cuda.is_available():
        return 'cuda'
    else:
        return 'cpu'

result = cpu()
print(result)
device = torch.device(result)
print(device)
```

```python
train_dir = r'C:\Users\lenovo\Desktop\graduation_project\data\new\Training Images'
test_dir = r'C:\Users\lenovo\Desktop\graduation_project\data\new\Testing Images'
val_dir = r'C:\Users\lenovo\Desktop\graduation_project\data\new\Validation Images'
excel_dir = r'C:\Users\lenovo\Desktop\graduation_project\data\raw\ODIR-5K\data.xlsx'

# 训练集
train_dataset = FundusDataset(train_dir ,excel_dir , is_training=True)

# 验证集
val_dataset = FundusDataset(val_dir ,excel_dir , is_training=False)

# 测试集
test_dataset = FundusDataset(test_dir ,excel_dir , is_training=False)
```

```python
# 创建DataLoader
train_loader = DataLoader(
    train_dataset,
    batch_size=config['batch_size'],
    shuffle=True,
    num_workers=4,  # 同时进行的进程数
    pin_memory=True,
    drop_last=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=config['batch_size'],
    shuffle=False,
    num_workers=4,
    pin_memory=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=config['batch_size'],
    shuffle=False,
    num_workers=4,
    pin_memory=True
)
```

```python
# 用已经预训练过的模型可以增加准确率
local_model_path = r'C:/Users/lenovo/Desktop/graduation_project/models/vit-base-patch16-224'

# 加载预训练的ViT模型
model = ViTForImageClassification.from_pretrained(
    local_model_path,
    num_labels=config['num_classes'],  # 指定分类数
    ignore_mismatched_sizes=True  # 允许模型权重尺寸不匹配时自动调整
)
model = model.to(device)
```

```python
import numpy as np

# 收集所有标签
all_labels = []
for i in range(len(train_dataset)):
    image, label, img_name = train_dataset[i]  # 接收3个值
    all_labels.append(label.numpy())

# 转换为numpy数组，shape: (样本数, 8)
all_labels = np.array(all_labels)

print("\n训练集标签分布:")
print("-" * 50)
print(f"{'疾病名称':<20} {'出现次数':<10} {'占比':<10} {'正负样本比'}")
print("-" * 50)

total_samples = len(all_labels)
for i, disease_name in enumerate(config['class_names']):
    positive_count = np.sum(all_labels[:, i] == 1)
    negative_count = total_samples - positive_count
    percentage = (positive_count / total_samples) * 100
    
    print(f"{disease_name:<20} {positive_count:<10} {percentage:>5.1f}%    {positive_count}:{negative_count}")

# 统计每个样本的疾病数量
disease_counts = np.sum(all_labels, axis=1)
print("\n" + "=" * 50)
print("每个样本的疾病数量分布:")
unique_counts, count_freq = np.unique(disease_counts, return_counts=True)
for num_diseases, freq in zip(unique_counts, count_freq):
    percentage = (freq / total_samples) * 100
    print(f"同时患 {int(num_diseases)} 种疾病: {freq} 张 ({percentage:.1f}%)")
```

```python
# 回归任务(预测连续值)：MSELoss,L1Loss,SmoothL1Loss
# 分类任务：1.多类分类(每个样本只属一类)：CrossEntropyLoss   2.多标签分类(每个样本可属多类)：BCEWithLogitsLoss

#数值更稳定，每个类别可独立判断
criterion = nn.BCEWithLogitsLoss()  # 损失函数

# 优化器
optimizer = torch.optim.AdamW(  #AdamW收敛快，准确率高，泛化性好
    model.parameters(),
    lr=config['learning_rate'],
    weight_decay=config['weight_decay']
)
```

```python
best_val = 0

for epoch in range(config['epochs']):
    model.train()
    train_loss = 0
    train_process = tqdm(train_loader,desc='Training')

for image, labels, img_name in train_process:
    image = image.to(device)
    labels = labels.to(device)
    
    optimizer.zero_grad()
    outputs = model(image)
    loss = criterion(outputs.logits,labels)
    loss.backward()
    optimizer.step()

    train_loss +=loss.item()
    train_process.set_postfix({'loss': loss.item()})
    avg_loss=train_loss / len(train_loader)
    
        
```

```python
from torch.utils.tensorboard import SummaryWriter
import torch
import torch.nn as nn

# 创建writer
writer = SummaryWriter('./logs')

for epoch in range(config['epochs']):
    # ========== 训练阶段 ==========
    model.train()
    train_loss = 0
    train_correct = 0
    train_total = 0
    
    for batch_idx, (images, labels, img_names) in enumerate(train_loader):
        images, labels = images.to(device), labels.to(device)
        
        outputs = model(images)
        loss = criterion(outputs.logits, labels)
        
        # 计算准确率
        probs = torch.sigmoid(outputs.logits)
        preds = (probs > 0.5).float()
        batch_correct = (preds == labels).all(dim=1).sum().item()
        batch_total = labels.size(0)
        
        loss.backward()
        optimizer.step()
        optimizer.zero_grad()
        
        train_loss += loss.item()
        train_correct += batch_correct
        train_total += batch_total
        
        # 记录每个batch的损失（每10个batch记录一次，避免太多点）
        if batch_idx % 10 == 0:
            global_step = epoch * len(train_loader) + batch_idx
            writer.add_scalar('Batch/Train_Loss', loss.item(), global_step)
            writer.add_scalar('Batch/Train_Accuracy', batch_correct/batch_total, global_step)
    
    # 计算epoch平均值
    avg_train_loss = train_loss / len(train_loader)
    train_acc = train_correct / train_total
    
    # ========== 验证阶段 ==========
    model.eval()
    val_loss = 0
    val_correct = 0
    val_total = 0
    all_preds = []
    all_labels = []
    
    with torch.no_grad():
        for images, labels, img_names in val_loader:
            images, labels = images.to(device), labels.to(device)
            
            outputs = model(images)
            loss = criterion(outputs.logits, labels)
            
            probs = torch.sigmoid(outputs.logits)
            preds = (probs > 0.5).float()
            
            val_loss += loss.item()
            val_correct += (preds == labels).all(dim=1).sum().item()
            val_total += labels.size(0)
            
            all_preds.append(preds.cpu().numpy())
            all_labels.append(labels.cpu().numpy())
    
    avg_val_loss = val_loss / len(val_loader)
    val_acc = val_correct / val_total
    
    # ========== 计算详细指标 ==========
    all_preds = np.vstack(all_preds)
    all_labels = np.vstack(all_labels)
    
    # 计算每个类别的F1
    from sklearn.metrics import f1_score
    macro_f1 = f1_score(all_labels, all_preds, average='macro', zero_division=0)
    micro_f1 = f1_score(all_labels, all_preds, average='micro', zero_division=0)
    per_class_f1 = f1_score(all_labels, all_preds, average=None, zero_division=0)
    
    # ========== 记录到TensorBoard ==========
    # 1. 损失曲线
    writer.add_scalar('Loss/Train', avg_train_loss, epoch)
    writer.add_scalar('Loss/Validation', avg_val_loss, epoch)
    
    # 2. 准确率曲线
    writer.add_scalar('Accuracy/Train', train_acc, epoch)
    writer.add_scalar('Accuracy/Validation', val_acc, epoch)
    
    # 3. F1分数
    writer.add_scalar('F1/Macro', macro_f1, epoch)
    writer.add_scalar('F1/Micro', micro_f1, epoch)
    
    # 4. 每个类别的F1
    for i, class_name in enumerate(config['class_names']):
        writer.add_scalar(f'F1_per_class/{class_name}', per_class_f1[i], epoch)
    
    # 5. 学习率
    current_lr = optimizer.param_groups[0]['lr']
    writer.add_scalar('Hyperparameters/Learning_Rate', current_lr, epoch)
    
    print(f"Epoch {epoch}: Train Loss={avg_train_loss:.4f}, Val Loss={avg_val_loss:.4f}, Macro F1={macro_f1:.4f}")
```

```python
# ==================== 2. 定义多标签评估函数 ====================
def multi_label_metrics(y_true, y_pred, threshold=0.5, class_names=None):
    """
    计算多标签分类的各项指标
    y_true: 真实标签，shape (n_samples, n_classes)
    y_pred: 预测标签（二值），shape (n_samples, n_classes)
    """
    
    # 1. 准确率 (Exact Match Ratio)
    exact_match = accuracy_score(y_true, y_pred)
    
    # 2. 每个类别的指标
    per_class_precision = precision_score(y_true, y_pred, average=None, zero_division=0)
    per_class_recall = recall_score(y_true, y_pred, average=None, zero_division=0)
    per_class_f1 = f1_score(y_true, y_pred, average=None, zero_division=0)
    
    # 3. 宏观平均
    macro_precision = precision_score(y_true, y_pred, average='macro', zero_division=0)
    macro_recall = recall_score(y_true, y_pred, average='macro', zero_division=0)
    macro_f1 = f1_score(y_true, y_pred, average='macro', zero_division=0)
    
    # 4. 微观平均
    micro_precision = precision_score(y_true, y_pred, average='micro', zero_division=0)
    micro_recall = recall_score(y_true, y_pred, average='micro', zero_division=0)
    micro_f1 = f1_score(y_true, y_pred, average='micro', zero_division=0)
    
    # 5. 汉明损失
    hamming_loss = np.mean(y_true != y_pred)
    
    # 打印每个类别的详细结果
    if class_names:
        print("\n每个类别的详细指标:")
        for i, name in enumerate(class_names):
            print(f"{name}:")
            print(f"  Precision: {per_class_precision[i]:.4f}")
            print(f"  Recall: {per_class_recall[i]:.4f}")
            print(f"  F1-score: {per_class_f1[i]:.4f}")
    
    return {
        'exact_match_ratio': exact_match,
        'macro_precision': macro_precision,
        'macro_recall': macro_recall,
        'macro_f1': macro_f1,
        'micro_precision': micro_precision,
        'micro_recall': micro_recall,
        'micro_f1': micro_f1,
        'hamming_loss': hamming_loss,
        'per_class_precision': per_class_precision,
        'per_class_recall': per_class_recall,
        'per_class_f1': per_class_f1
    }
```

```python
from tqdm import tqdm 

best_val_f1 = 0

for epoch in range(config['epochs']):
    print(f"\n{'='*50}")
    print(f"Epoch {epoch+1}/{config['epochs']}")
    print(f"{'='*50}")
    
    # ========== 训练阶段 ==========
    model.train()
    train_loss = 0
    
    # 使用tqdm创建进度条
    train_bar = tqdm(train_loader, desc='Training')
    
    for batch_idx, (images, labels, img_names) in enumerate(train_bar):
        images, labels = images.to(device), labels.to(device)
        
        # 前向传播
        outputs = model(images)
        loss = criterion(outputs.logits, labels)
        
        # 反向传播
        loss.backward()
        optimizer.step()
        optimizer.zero_grad()
        
        # 累计损失
        train_loss += loss.item()
        
        # 在进度条上显示当前loss
        train_bar.set_postfix({'loss': loss.item()})
    
    avg_train_loss = train_loss / len(train_loader)
    print(f"\n训练损失: {avg_train_loss:.4f}")
```
